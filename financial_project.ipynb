```python
import fitz  # PyMuPDF
import pandas as pd
import re

pdf_path = r"D:\guvi final project\personal_financial_transcactions.pdf"

transactions = []

# Open PDF with fitz
doc = fitz.open(pdf_path)
for page in doc:
    text = page.get_text("text")
    lines = text.split("\n")
    
    for line in lines:
        # Match: date time amount direction counterparty...ref_id category
        match = re.match(r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+(\d+)\s+(debit|credit)\s+(.*?)(UTR\d+)([A-Za-z]+)", line)
        if match:
            date, time, amount, direction, counterparty_desc, ref_id, category = match.groups()
            
            # Split counterparty and description
            if "Payment related to" in counterparty_desc:
                parts = counterparty_desc.split("Payment related to")
                counterparty = parts[0].strip()
                description = "Payment related to " + parts[1].strip()
            else:
                counterparty = counterparty_desc.strip()
                description = ""
            
            transactions.append([date, time, amount, direction, counterparty, description, ref_id, category])

doc.close()

# Convert to DataFrame
columns = ["date", "time", "amount", "direction", "counterparty", "description", "ref_id", "category"]
df = pd.DataFrame(transactions, columns=columns)

# Convert numeric columns
df["amount"] = pd.to_numeric(df["amount"], errors="coerce")

# Save as CSV
df.to_csv("transactions.csv", index=False)

print("Data extracted successfully!")
print(df.head(10))
```

```python
# robust_extract_transactions.py
# Run this in the same environment where you run previous code.
# Make sure you have fitz (PyMuPDF). If PDF is image-only, you'll need pdf2image + pytesseract.

import fitz            # pip install pymupdf
import re
import pandas as pd
import os
import sys

pdf_path = r"D:\guvi final project\personal_financial_transcactions.pdf"  # adjust if different

def print_raw_lines(pdf_path, pages_to_show=2, lines_to_show=30):
    """Print raw text lines so you can inspect actual formatting."""
    doc = fitz.open(pdf_path)
    for page_num in range(min(len(doc), pages_to_show)):
        page = doc[page_num]
        text = page.get_text("text")
        lines = text.splitlines()
        print(f"\n--- Page {page_num+1} raw lines (first {lines_to_show}) ---")
        for i, line in enumerate(lines[:lines_to_show]):
            # show repr to reveal hidden characters/spaces
            print(i+1, repr(line))
    doc.close()

def extract_with_patterns(pdf_path, patterns):
    """Try multiple regex patterns to extract rows. Return list of rows."""
    doc = fitz.open(pdf_path)
    rows = []
    for page in doc:
        text = page.get_text("text")
        if not text or text.strip()=="":
            continue
        for line in text.splitlines():
            for pat in patterns:
                m = re.search(pat, line)
                if m:
                    rows.append(m.groups())
                    break
    doc.close()
    return rows

def try_ocr(pdf_path):
    """Fallback: OCR pages to text and return lines (requires pdf2image & pytesseract)."""
    try:
        from pdf2image import convert_from_path  # pip install pdf2image
        import pytesseract                        # pip install pytesseract
    except Exception as e:
        print("OCR fallback unavailable (pdf2image/pytesseract not installed).", e)
        return []

    print("Starting OCR fallback (slower). Make sure Tesseract is installed on your system.")
    pages = convert_from_path(pdf_path, dpi=200)
    ocr_texts = []
    for i, page_image in enumerate(pages):
        txt = pytesseract.image_to_string(page_image)
        lines = txt.splitlines()
        print(f"OCR page {i+1} first 10 lines: ")
        for l in lines[:10]:
            print(repr(l))
        ocr_texts.append("\n".join(lines))
    return ocr_texts

# --- Step A: show raw lines so you (and we) know the format ---
print("1) Printing raw lines from the PDF. Inspect these to see exact formatting:")
if not os.path.exists(pdf_path):
    print("ERROR: PDF not found at", pdf_path)
    sys.exit(1)
print_raw_lines(pdf_path, pages_to_show=2, lines_to_show=40)

# --- Step B: try multiple patterns (covers common line formats) ---
# We'll attempt patterns that capture:
#   date time amount direction counterparty (Payment related to ...)? UTRxxx category
patterns = [
    # pattern 1: same as earlier, but made flexible about spacing and optional description text
    r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+(\d+)\s+(debit|credit)\s+(.*?)\s*(?:Payment related to\s*(.*?))?\s*(UTR\d+)\s*([A-Za-z]+)",
    # pattern 2: sometimes description and UTR are glued (no space between UTR and category)
    r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+(\d+)\s+(debit|credit)\s+(.*?)(UTR\d+)([A-Za-z]+)",
    # pattern 3: minimal greedy split: date time amount dir rest category at end
    r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+(\d+)\s+(debit|credit)\s+(.*)\s+(UTR\d+)\s*([A-Za-z]+)$",
    # pattern 4: if category is absent or numeric; capture everything to inspect
    r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+(\d+)\s+(debit|credit)\s+(.*)",
]

print("\n2) Trying multiple regex patterns to extract transactions...")
rows = extract_with_patterns(pdf_path, patterns)

# If none found, report and suggest OCR
if not rows:
    print("\nNo matches found with text extraction. Possible causes:")
    print("- PDF may be image-based (scanned) so text extraction returns empty or garbage.")
    print("- The actual line format differs; check the raw lines above for differences.")
    # Try OCR fallback
    ocr_texts = try_ocr(pdf_path)
    ocr_rows = []
    if ocr_texts:
        # try the same patterns over OCR text
        for ocr_text in ocr_texts:
            for line in ocr_text.splitlines():
                for pat in patterns:
                    m = re.search(pat, line)
                    if m:
                        ocr_rows.append(m.groups())
                        break
    if ocr_rows:
        rows = ocr_rows
        print(f"\nOCR-based extraction found {len(rows)} matches.")
    else:
        print("\nAfter OCR, still no matches. Paste here a few raw lines (use print_raw_lines output) and I'll adapt the pattern.")
        sys.exit(0)

# --- Step C: normalize extracted tuples into DataFrame columns ---
print(f"\nFound {len(rows)} raw regex matches. Converting to dataframe...")
normalized = []
for r in rows:
    # Different patterns produce different tuple shapes — handle them.
    # We'll try to safely map fields: date,time,amount,direction,counterparty,description,ref_id,category
    date = time = amount = direction = counterparty = description = ref_id = category = None
    if len(r) >= 8:
        date, time, amount, direction, cp1, desc1, ref_id, category = r[:8]
        # sometimes cp1 contains 'merchant' and desc1 empty. reassemble:
        if desc1 is None or desc1.strip()=="":
            # if cp1 contains "Payment related to", split it
            if "Payment related to" in cp1:
                parts = cp1.split("Payment related to")
                counterparty = parts[0].strip()
                description = "Payment related to " + parts[1].strip()
            else:
                counterparty = cp1.strip()
                description = ""
        else:
            counterparty = cp1.strip()
            description = desc1.strip()
    elif len(r) == 7:
        date, time, amount, direction, cp, ref_id, category = r
        counterparty = cp.strip()
        description = ""
    elif len(r) == 5:
        date, time, amount, direction, rest = r
        # try to split rest into counterparty/ref/category heuristically
        # e.g., 'Flipkart Payment related to FlipkartUTR100000Shopping'
        m = re.search(r"(.*?)(UTR\d+)([A-Za-z]+)$", rest)
        if m:
            counterparty_desc = m.group(1).strip()
            ref_id = m.group(2)
            category = m.group(3)
            if "Payment related to" in counterparty_desc:
                parts = counterparty_desc.split("Payment related to")
                counterparty = parts[0].strip()
                description = "Payment related to " + parts[1].strip()
            else:
                # split first token as counterparty
                tokens = counterparty_desc.split()
                counterparty = tokens[0]
                description = counterparty_desc
        else:
            counterparty = rest.strip()
            description = ""
    else:
        # unknown shape: stash the joined values in description to inspect later
        joined = " | ".join([str(x) for x in r])
        date = r[0] if len(r)>0 else None
        counterparty = joined
        description = joined

    normalized.append({
        "date": date,
        "time": time,
        "amount": amount,
        "direction": direction,
        "counterparty": counterparty,
        "description": description,
        "ref_id": ref_id,
        "category": category
    })

df = pd.DataFrame(normalized, columns=["date","time","amount","direction","counterparty","description","ref_id","category"])

# clean numeric
df["amount"] = pd.to_numeric(df["amount"], errors="coerce")

print("\nDataFrame shape:", df.shape)
print("First 10 rows (for inspection):")
print(df.head(10))

# Save CSV
out_csv = "transactions_extracted.csv"
df.to_csv(out_csv, index=False)
print(f"\nSaved extracted transactions to {out_csv}")

# Helpful tip: if df is empty, open the saved CSV to see intermediate content.
if df.empty:
    print("\nDataFrame is still empty after all attempts. Two likely reasons:")
    print("1) The PDF text format is quite different — check the raw lines printed above.")
    print("2) The file is image-based and OCR failed or not installed. Install tesseract and pytesseract and try again.")
else:
    print("\nSUCCESS: dataset extraction complete (or partial). You can now proceed to cleaning & analysis.")
```

```python
#today night work

#!/usr/bin/env python3
"""
transaction_extractor.py

Extracts structured transaction data from a bank statements PDF.
If text extraction fails or PDF is image-only, uses OCR via pdf2image + pytesseract.

Outputs a CSV file with columns:
  date, time, amount, direction, counterparty, description, ref_id, category

Usage:
    python transaction_extractor.py <path_to_pdf> [--poppler-path <poppler_bin_path>]

Dependencies:
  - pymupdf (fitz): pip install pymupdf
  - pdf2image: pip install pdf2image
  - pytesseract: pip install pytesseract
  - pandas
  - re, os, sys, logging

Ensure Poppler is installed (on Windows extract poppler binaries, add poppler/bin to PATH or pass via --poppler-path).
Ensure Tesseract OCR is installed and pytesseract is configured.
"""

import fitz  # PyMuPDF
import re
import pandas as pd
import os
import sys
import argparse
import logging

def setup_logging():
    # Configure logging
    logging.basicConfig(
        level=logging.INFO,
        format='%(asctime)s [%(levelname)s] %(message)s',
        handlers=[
            logging.StreamHandler(sys.stdout)
        ]
    )

def print_raw_lines(pdf_path, pages_to_show=2, lines_to_show=30):
    """Print raw text lines so you can inspect actual formatting."""
    logging.info(f"Opening PDF for raw line inspection: {pdf_path}")
    try:
        doc = fitz.open(pdf_path)
    except Exception as e:
        logging.error(f"Could not open PDF: {e}")
        return

    for page_num in range(min(len(doc), pages_to_show)):
        page = doc[page_num]
        text = page.get_text("text")
        lines = text.splitlines()
        logging.info(f"\n--- Page {page_num+1} raw lines (first {lines_to_show}) ---")
        for i, line in enumerate(lines[:lines_to_show]):
            # show repr to reveal hidden characters/spaces
            logging.info(f"{i+1}: {repr(line)}")
    doc.close()

def extract_with_patterns_from_text(text, patterns):
    """Try multiple regex patterns over a block of text. Return list of matched groups."""
    rows = []
    for line in text.splitlines():
        for pat in patterns:
            m = re.search(pat, line)
            if m:
                rows.append(m.groups())
                break
    return rows

def extract_with_patterns(pdf_path, patterns):
    """Extract using text extraction (non-OCR)."""
    logging.info("Attempting text extraction (non-OCR)...")
    try:
        doc = fitz.open(pdf_path)
    except Exception as e:
        logging.error(f"Error opening PDF: {e}")
        return []

    rows = []
    for page in doc:
        try:
            text = page.get_text("text")
        except Exception as e:
            logging.warning(f"Error extracting text from a page: {e}")
            continue

        if not text or text.strip() == "":
            continue

        page_rows = extract_with_patterns_from_text(text, patterns)
        if page_rows:
            rows.extend(page_rows)

    doc.close()
    logging.info(f"Text extraction found {len(rows)} rows using patterns.")
    return rows

def try_ocr(pdf_path, patterns, poppler_path=None, dpi=200):
    """Fallback: OCR pages to text and extract using patterns."""
    logging.info("Starting OCR fallback...")
    try:
        from pdf2image import convert_from_path
        import pytesseract
    except ImportError as e:
        logging.error("OCR fallback unavailable. Make sure pdf2image and pytesseract are installed.")
        return []

    # If poppler_path is provided, pass it
    convert_kwargs = {}
    if poppler_path:
        convert_kwargs['poppler_path'] = poppler_path

    try:
        pages = convert_from_path(pdf_path, dpi=dpi, **convert_kwargs)
    except Exception as e:
        logging.error(f"Could not convert PDF to images for OCR: {e}")
        return []

    ocr_rows = []
    for i, page_image in enumerate(pages):
        try:
            txt = pytesseract.image_to_string(page_image)
        except Exception as e:
            logging.warning(f"Error during OCR on page {i+1}: {e}")
            continue

        logging.info(f"OCR page {i+1} first 10 lines: ")
        lines = txt.splitlines()
        for l in lines[:10]:
            logging.info(repr(l))

        # extract from OCR text
        page_rows = extract_with_patterns_from_text(txt, patterns)
        if page_rows:
            ocr_rows.extend(page_rows)

    logging.info(f"OCR extraction found {len(ocr_rows)} rows using patterns.")
    return ocr_rows

def normalize_rows(rows):
    """Normalize extracted tuples into dicts with known keys."""
    normalized = []
    for r in rows:
        # initialize all fields to None
        date = time = amount = direction = counterparty = description = ref_id = category = None

        # Different patterns produce different tuple lengths
        if len(r) >= 8:
            date, time, amount, direction, cp1, desc1, ref_id, category = r[:8]
            counterparty = cp1.strip() if cp1 else None
            description = desc1.strip() if desc1 else ""
        elif len(r) == 7:
            date, time, amount, direction, cp, ref_id, category = r
            counterparty = cp.strip()
            description = ""
        elif len(r) == 5:
            date, time, amount, direction, rest = r
            # attempt to split rest into cp / ref_id / category
            m = re.search(r"(.*?)(UTR\d+)([A-Za-z]+)$", rest)
            if m:
                counterparty_desc = m.group(1).strip()
                ref_id = m.group(2)
                category = m.group(3)
                if "Payment related to" in counterparty_desc:
                    parts = counterparty_desc.split("Payment related to")
                    counterparty = parts[0].strip()
                    description = "Payment related to " + parts[1].strip()
                else:
                    # simple heuristic
                    counterparty = counterparty_desc
                    description = ""
            else:
                counterparty = rest.strip()
                description = ""
        else:
            # fallback: join all as description
            joined = " | ".join([str(x) for x in r])
            date = r[0] if len(r) > 0 else None
            counterparty = joined
            description = joined

        normalized.append({
            "date": date,
            "time": time,
            "amount": amount,
            "direction": direction,
            "counterparty": counterparty,
            "description": description,
            "ref_id": ref_id,
            "category": category
        })
    return normalized

def parse_args():
    parser = argparse.ArgumentParser(description="Extract transactions from PDF (with OCR fallback)")
    parser.add_argument("pdf_path", help="Path to the PDF file containing transactions")
    parser.add_argument("--poppler-path", dest="poppler_path", default=None,
                        help="Path to Poppler bin folder (if not in system PATH)")
    parser.add_argument("--output", "-o", dest="output_csv", default="transactions_extracted.csv",
                        help="Name/path of output CSV file")
    parser.add_argument("--debug-inspect", dest="debug_inspect", action="store_true",
                        help="Print raw lines from PDF for inspection")
    parser.add_argument("--dpi", dest="dpi", type=int, default=200,
                        help="DPI resolution for OCR conversion (higher gives better OCR but slower)")
    return parser.parse_args()

def main():
    setup_logging()
    args = parse_args()

    pdf_path = args.pdf_path
    poppler_path = args.poppler_path
    output_csv = args.output_csv
    dpi = args.dpi

    # Check PDF file exists
    if not os.path.isfile(pdf_path):
        logging.error(f"PDF file not found: {pdf_path}")
        sys.exit(1)

    # Optional: inspect raw text to see formatting
    if args.debug_inspect:
        print_raw_lines(pdf_path, pages_to_show=2, lines_to_show=40)

    # Define regex patterns (modify as needed for your document format)
    patterns = [
        # Pattern 1: date, time, amount, direction, counterparty, optional "Payment related to", UTR, category
        r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+([\d,]+(?:\.\d+)?)\s+(debit|credit)\s+(.*?)\s*(?:Payment related to\s*(.*?))?\s*(UTR\d+)\s*([A-Za-z]+)",
        # Pattern 2: description and UTR glued
        r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+([\d,]+(?:\.\d+)?)\s+(debit|credit)\s+(.*?)(UTR\d+)([A-Za-z]+)",
        # Pattern 3: category at end
        r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+([\d,]+(?:\.\d+)?)\s+(debit|credit)\s+(.*)\s+(UTR\d+)\s*([A-Za-z]+)$",
        # Pattern 4: minimal (no category/UTR)
        r"(\d{2}-\d{2}-\d{4})\s+(\d{2}:\d{2})\s+([\d,]+(?:\.\d+)?)\s+(debit|credit)\s+(.*)",
    ]

    # Step 1: Try normal text extraction
    rows = extract_with_patterns(pdf_path, patterns)

    # Step 2: If not enough matches, fallback to OCR
    if not rows or len(rows) < 5:
        logging.info("Text extraction gave insufficient results (rows < 5). Trying OCR fallback.")
        ocr_rows = try_ocr(pdf_path, patterns, poppler_path=poppler_path, dpi=dpi)
        if ocr_rows:
            rows = ocr_rows
        else:
            logging.error("After OCR fallback, still no matches found.")
            # Optionally exit or continue with whatever found
            if not rows:
                logging.error("No data to write. Exiting.")
                sys.exit(1)

    # Step 3: Normalize
    normalized = normalize_rows(rows)
    df = pd.DataFrame(normalized, columns=["date", "time", "amount", "direction",
                                           "counterparty", "description", "ref_id", "category"])

    # Clean amount: remove commas, convert to float
    def clean_amount(x):
        if x is None:
            return None
        try:
            # Remove commas
            s = str(x).replace(",", "")
            # Maybe it's integer or float
            return float(s)
        except ValueError:
            return None

    df["amount"] = df["amount"].apply(clean_amount)

    # Optional: drop rows with missing essential fields
    df_clean = df.dropna(subset=["date", "time", "amount", "direction"])

    # Save CSV
    logging.info(f"Saving output to: {output_csv}")
    try:
        df_clean.to_csv(output_csv, index=False)
    except Exception as e:
        logging.error(f"Error saving CSV: {e}")
        sys.exit(1)

    logging.info(f"Done. Extracted {len(df_clean)} transactions.")

if __name__ == "__main__":
    main()
```

```python
import re
from datetime import datetime
import pandas as pd
import pdfplumber
# optional: import fitz  # PyMuPDF if you prefer

DATE_FORMATS = [
    "%d-%m-%Y %H:%M", "%d-%m-%Y %H:%M:%S",
    "%d-%m-%Y", "%Y-%m-%d %H:%M", "%d/%m/%Y", "%d-%m-%Y %H:%M"
]

# Basic keyword -> category mapping (extendable)
CATEGORY_KEYWORDS = {
    "Shopping": ["flipkart", "amazon", "myntra", "snapdeal"],
    "Food": ["zomato", "swiggy", "dominos", "pizza", "restaurant", "dine"],
    "Transport": ["ola", "uber", "rapido", "taxi"],
    "Bills": ["airtel", "bsnl", "tneb", "electricity", "bill", "vodafone"],
    "Rent": ["house owner", "flat rent", "rent"],
    "Salary": ["company", "salary", "payroll"],
    "Investment": ["sbi securities", "mutual fund", "sip", "investment"],
    "Entertainment": ["hotstar", "netflix", "bookmyshow", "hotstar", "hotstar"],
    "Groceries": ["bigbasket", "blinkit", "zepto"],
    "Transfer": ["self account", "friend transfer"],
}

def guess_category(text):
    if pd.isna(text):
        return "Unknown"
    t = text.lower()
    for cat, keywords in CATEGORY_KEYWORDS.items():
        for kw in keywords:
            if kw in t:
                return cat
    # fallback heuristics
    if "credit" in t and "salary" in t: return "Salary"
    if "transfer" in t: return "Transfer"
    return "Other"

def parse_table_like_text(text):
    """
    If the PDF contains rows like:
    date time amount direction counterpart description ref_id category
    this function will attempt to parse lines into columns.
    """
    rows = []
    for line in text.splitlines():
        line = line.strip()
        if not line:
            continue
        # try to match: DATE TIME AMOUNT DIRECTION MERCHANT rest...
        # Many PDF exports are space-separated with no consistent delimiter; use regex
        # Example line (from your dataset): "02-09-2024 19:02 17734 debit Flipkart Payment related to FlipkartUTR100000Shopping"
        m = re.match(r"(\d{2}[-/]\d{2}[-/]\d{4})\s+(\d{1,2}:\d{2})\s+(\d+)\s+(\w+)\s+(.+)", line)
        if m:
            date_s, time_s, amount_s, direction, rest = m.groups()
            # attempt to split rest into merchant + description + maybe UTR + category
            # UTR and category may be jammed; try to find UTR like 'UTR\d+'
            utr = None
            utr_m = re.search(r"(UTR\d+)", rest)
            if utr_m:
                utr = utr_m.group(1)
                rest = rest.replace(utr, "").strip()
            # very simple merchant extraction: first token up to next capitalized word or 'Payment'
            parts = rest.split("Payment related to")
            if len(parts) == 2:
                merchant = parts[0].strip()
                description = parts[1].strip()
            else:
                tokens = rest.split()
                merchant = tokens[0]
                description = " ".join(tokens[1:]) if len(tokens) > 1 else ""
            rows.append({
                "date": date_s,
                "time": time_s,
                "amount": float(amount_s),
                "direction": direction,
                "merchant": merchant.strip(),
                "description": description.strip(),
                "utr": utr
            })
        else:
            # fallback: try splitting by multiple spaces
            toks = re.split(r"\s{2,}", line)
            if len(toks) >= 4:
                rows.append({
                    "date": toks[0],
                    "time": toks[1] if len(toks) > 1 else "",
                    "amount": float(re.sub(r"[^\d.]", "", toks[2])),
                    "direction": toks[3] if len(toks) > 3 else "",
                    "merchant": toks[4] if len(toks) > 4 else "",
                    "description": " ".join(toks[5:]) if len(toks) > 5 else ""
                })
            else:
                # ignore junk
                continue
    df = pd.DataFrame(rows)
    return df

def extract_text_from_pdf(path):
    """
    Use pdfplumber to extract text from all pages.
    Returns concatenated text.
    """
    texts = []
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            txt = page.extract_text()
            if txt:
                texts.append(txt)
    return "\n".join(texts)

def parse_pdf_to_df(path):
    """
    Master parser: extract text and try to parse table-like lines.
    """
    text = extract_text_from_pdf(path)
    # quick attempt: see if the file already contains a header 'date time amount'
    if re.search(r"\bdate\b.*\btime\b.*\bamount\b", text, re.I):
        # try to directly parse lines
        df = parse_table_like_text(text)
    else:
        # naive: split lines and attempt to parse
        df = parse_table_like_text(text)
    # normalize
    if df.empty:
        return df
    # try to combine date+time into datetime
    def parse_dt(row):
        s = f"{row.get('date','')} {row.get('time','')}".strip()
        for fmt in DATE_FORMATS:
            try:
                return datetime.strptime(s, fmt)
            except Exception:
                continue
        # fallback: try dd-mm-YYYY
        try:
            return datetime.strptime(row.get('date',''), "%d-%m-%Y")
        except Exception:
            return pd.NaT
    df['datetime'] = df.apply(parse_dt, axis=1)
    # fill merchant/description
    df['merchant'] = df['merchant'].astype(str).str.strip()
    df['description'] = df['description'].astype(str).str.strip()
    df['category'] = df.apply(lambda r: guess_category(" ".join([str(r['merchant']), str(r['description']), str(r.get('direction',''))])), axis=1)
    # normalize amount sign by direction
    df['amount_signed'] = df.apply(lambda r: -r['amount'] if str(r.get('direction','')).lower() in ['debit','dr'] else r['amount'], axis=1)
    # final columns
    df = df[['datetime','date','time','amount','amount_signed','direction','merchant','description','utr','category']]
    return df
```

```python
import os
import openai

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY", "")
if OPENAI_API_KEY:
    openai.api_key = OPENAI_API_KEY

def get_recommendations(transactions_df, user_profile=None, model="gpt-4o-mini" ):
    """
    Build a concise prompt describing spending patterns and ask the LLM for advice.
    Returns text.
    """
    # build summary
    total_spent = transactions_df[transactions_df['amount_signed'] < 0]['amount'].sum()
    total_income = transactions_df[transactions_df['amount_signed'] > 0]['amount'].sum()
    top_categories = transactions_df.groupby('category')['amount'].sum().abs().sort_values(ascending=False).head(5).to_dict()
    recent = transactions_df.sort_values('datetime', ascending=False).head(10)[['datetime','merchant','amount','category']].to_dict(orient='records')

    prompt = f"""
You are a helpful personal finance assistant. A user gave their transaction data. Summary:
- total income: {total_income:.2f}
- total spending: {total_spent:.2f}
- top categories: {top_categories}
Provide: (1) a one-paragraph summary of their spending behavior, (2) 5 actionable recommendations to reduce unnecessary expenses and improve savings, (3) a suggested monthly budget split by category (percentages), and (4) any suspicious or wasteful transactions to watch out for based on top categories.

Recent transactions (latest 10): {recent}
Respond in plain text, numbered lists where appropriate.
"""
    # Use OpenAI (if key present)
    if OPENAI_API_KEY:
        try:
            resp = openai.ChatCompletion.create(
                model="gpt-4o-mini",  # change as per availability
                messages=[{"role":"user","content":prompt}],
                max_tokens=600,
                temperature=0.2
            )
            return resp['choices'][0]['message']['content'].strip()
        except Exception as e:
            return f"LLM call failed: {e}\n\nPrompt used:\n{prompt}"
    else:
        # fallback simple rule-based stub
        recs = ["(No API key set) Install OpenAI key to get personalized LLM recommendations.",
                "Rule: Reduce top discretionary category by 15% next month.",
                "Rule: Set auto-transfer to savings 10% of income.",
                "Rule: Review subscriptions (Entertainment) and cancel unused ones.",
                "Rule: Track weekly groceries & aim for 10% saving."]
        return "LLM not configured. Suggestions:\n" + "\n".join(f"- {r}" for r in recs)
```

```python
# app.py — Personal UPI Usage & Financial Analyzer (single-file)
# Run: streamlit run app.py

import re
import os
import tempfile
from datetime import datetime
from io import BytesIO
import base64

import pandas as pd
import pdfplumber

import streamlit as st
import plotly.express as px

# Optional: enable OpenAI LLM recommendations (set OPENAI_API_KEY as env var)
try:
    import openai
except Exception:
    openai = None

# ----------------------------
# Config & helper data
# ----------------------------
st.set_page_config(page_title="Personal UPI Analyzer", layout="wide")

DATE_FORMATS = [
    "%d-%m-%Y %H:%M", "%d-%m-%Y %H:%M:%S", "%d-%m-%Y",
    "%Y-%m-%d %H:%M", "%Y-%m-%d", "%d/%m/%Y",
    "%d-%m-%Y %H:%M"
]

CATEGORY_KEYWORDS = {
    "Shopping": ["flipkart", "amazon", "myntra", "snapdeal"],
    "Food": ["zomato", "swiggy", "dominos", "pizza", "restaurant", "dine"],
    "Transport": ["ola", "uber", "rapido", "taxi"],
    "Bills": ["airtel", "bsnl", "tneb", "electricity", "bill", "vodafone"],
    "Rent": ["house owner", "flat rent", "rent"],
    "Salary": ["company", "salary", "payroll"],
    "Investment": ["sbi securities", "mutual fund", "sip", "investment"],
    "Entertainment": ["hotstar", "netflix", "bookmyshow"],
    "Groceries": ["bigbasket", "blinkit", "zepto"],
    "Transfer": ["self account", "friend transfer", "transfer"],
}

DIRECTION_DEBIT = {"debit", "dr", "withdrawal", "sent", "paid"}
DIRECTION_CREDIT = {"credit", "cr", "received", "deposit", "refund"}

# ----------------------------
# Parsing helpers
# ----------------------------
def clean_amount_str(s):
    if s is None:
        return None
    s = str(s)
    # remove non-digit except dot and minus
    s = s.replace(",", "")
    m = re.search(r"-?\d+(\.\d+)?", s)
    return float(m.group(0)) if m else None

def guess_category(text):
    if not text or pd.isna(text):
        return "Unknown"
    t = str(text).lower()
    for cat, keywords in CATEGORY_KEYWORDS.items():
        for kw in keywords:
            if kw in t:
                return cat
    if "salary" in t or "payroll" in t: return "Salary"
    if "transfer" in t: return "Transfer"
    return "Other"

def extract_text_from_pdf_bytes(bytes_data):
    texts = []
    try:
        with pdfplumber.open(BytesIO(bytes_data)) as pdf:
            for page in pdf.pages:
                txt = page.extract_text()
                if txt:
                    texts.append(txt)
    except Exception as e:
        # file might be corrupt or scanned image; return empty
        st.warning(f"pdfplumber failed to read PDF: {e}")
    return "\n".join(texts)

def parse_table_like_text(text):
    """
    Flexible parser: tries several patterns and fallback heuristics.
    Returns dataframe with columns: date,time,amount,direction,merchant,description,utr
    """
    rows = []
    if not text:
        return pd.DataFrame(rows)

    for line in text.splitlines():
        line = line.strip()
        if not line:
            continue

        # normalize multiple spaces
        line_norm = re.sub(r"\s{2,}", "  ", line)

        # Pattern 1: dd-mm-yyyy [HH:MM] amount direction merchant ... (common)
        m = re.match(r"^(\d{1,2}[-/]\d{1,2}[-/]\d{4})\s+(\d{1,2}:\d{2})?\s+([-\d,\.]+)\s+([A-Za-z]+)\s+(.+)$", line_norm)
        if m:
            date_s, time_s, amount_s, direction, rest = m.groups()
            amt = clean_amount_str(amount_s)
            utr = None
            utr_m = re.search(r"(UTR[:\s]?[\w\d-]+)", rest, re.I)
            if utr_m:
                utr = utr_m.group(1)
                rest = rest.replace(utr, "").strip()
            # merchant heuristics: up to known keywords or 'Payment' words
            parts = re.split(r"\bPayment related to\b|\bfor\b|\b-+\b", rest, flags=re.I)
            merchant = parts[0].strip() if parts else ""
            description = parts[1].strip() if len(parts) > 1 else rest.replace(merchant, "").strip()
            rows.append({
                "date": date_s,
                "time": time_s if time_s else "",
                "amount": float(amt) if amt is not None else None,
                "direction": direction,
                "merchant": merchant,
                "description": description,
                "utr": utr
            })
            continue

        # Pattern 2: lines with 'debit'/'credit' anywhere and amount nearby
        if re.search(r"\b(debit|credit|dr|cr)\b", line, re.I) and re.search(r"\d[\d,\.]+\b", line):
            amt = clean_amount_str(line)
            dir_m = re.search(r"\b(debit|credit|dr|cr|paid|sent|received|deposit)\b", line, re.I)
            direction = dir_m.group(1) if dir_m else ""
            # attempt to extract date from line
            date_m = re.search(r"(\d{1,2}[-/]\d{1,2}[-/]\d{4})", line)
            time_m = re.search(r"(\d{1,2}:\d{2})", line)
            rest = line
            if date_m:
                rest = rest.replace(date_m.group(0), "")
            if time_m:
                rest = rest.replace(time_m.group(0), "")
            # best-effort merchant = first word chunk
            merchant = rest.split()[0] if rest.split() else ""
            description = rest.strip()
            rows.append({
                "date": date_m.group(0) if date_m else "",
                "time": time_m.group(0) if time_m else "",
                "amount": float(amt) if amt is not None else None,
                "direction": direction,
                "merchant": merchant,
                "description": description,
                "utr": None
            })
            continue

        # Fallback: try splitting by two or more spaces (table-like)
        toks = re.split(r"\s{2,}", line)
        if len(toks) >= 4:
            amt = clean_amount_str(toks[2])
            rows.append({
                "date": toks[0],
                "time": toks[1] if len(toks) > 1 else "",
                "amount": float(amt) if amt is not None else None,
                "direction": toks[3] if len(toks) > 3 else "",
                "merchant": toks[4] if len(toks) > 4 else "",
                "description": " ".join(toks[5:]) if len(toks) > 5 else ""
            })
            continue

        # otherwise: ignore noisy line
        continue

    df = pd.DataFrame(rows)
    return df

# ----------------------------
# Master parse + normalization
# ----------------------------
def parse_pdf_to_df_from_bytes(bytes_data):
    text = extract_text_from_pdf_bytes(bytes_data)
    df = parse_table_like_text(text)
    if df.empty:
        return df

    # standardize columns
    if 'amount' in df.columns:
        df['amount'] = pd.to_numeric(df['amount'], errors='coerce').fillna(0.0)
    else:
        df['amount'] = 0.0

    # combine date+time -> datetime
    def parse_dt(row):
        s = f"{row.get('date','')} {row.get('time','')}".strip()
        for fmt in DATE_FORMATS:
            try:
                return datetime.strptime(s, fmt)
            except Exception:
                continue
        # fallback: try date only heuristics
        for fmt in ["%d-%m-%Y", "%d/%m/%Y", "%Y-%m-%d"]:
            try:
                return datetime.strptime(row.get('date','') or '', fmt)
            except Exception:
                continue
        return pd.NaT

    df['datetime'] = df.apply(parse_dt, axis=1)

    # strip/normalize merchant & description
    for c in ['merchant','description','direction','utr']:
        if c in df.columns:
            df[c] = df[c].astype(str).str.strip().replace({'nan': ''})

    # category
    df['category'] = df.apply(lambda r: guess_category(" ".join([str(r.get('merchant','')), str(r.get('description','')), str(r.get('direction',''))])), axis=1)

    # amount_signed: debits negative
    def signed_amount(r):
        d = str(r.get('direction','')).lower()
        amt = r.get('amount', 0.0) or 0.0
        if any(x in d for x in DIRECTION_DEBIT):
            return -abs(amt)
        if any(x in d for x in DIRECTION_CREDIT):
            return abs(amt)
        # fallback heuristics using keywords in description
        desc = str(r.get('description','')).lower()
        if "received" in desc or "deposit" in desc:
            return abs(amt)
        return -abs(amt)  # assume spending if unclear

    df['amount_signed'] = df.apply(signed_amount, axis=1)

    # ensure final columns order
    final_cols = ['datetime','date','time','amount','amount_signed','direction','merchant','description','utr','category']
    for c in final_cols:
        if c not in df.columns:
            df[c] = ""
    df = df[final_cols]
    return df

# ----------------------------
# LLM recommendations helper
# ----------------------------
#api vaiable value
OPENAI_API_KEY = "sk-xxxxxxxxxxxxxxxxxxxxxxxx"
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY", "")
if OPENAI_API_KEY and openai:
    openai.api_key = OPENAI_API_KEY

def get_recommendations(transactions_df):
    # fallback: if no valid rows, return helpful message
    if transactions_df.empty:
        return "No transactions available to analyze."

    total_spent = transactions_df[transactions_df['amount_signed'] < 0]['amount'].abs().sum()
    total_income = transactions_df[transactions_df['amount_signed'] > 0]['amount'].sum()
    top_categories = transactions_df.groupby('category')['amount_signed'].sum().abs().sort_values(ascending=False).head(5).to_dict()
    recent = transactions_df.sort_values('datetime', ascending=False).head(10)[['datetime','merchant','amount_signed','category']].to_dict(orient='records')

    prompt = f"""
You are a helpful personal finance assistant. Summarize this user's spending data and give actionable recommendations.
Total income: {total_income:.2f}
Total spending: {total_spent:.2f}
Top categories: {top_categories}
Recent transactions: {recent}

Provide:
1) One-paragraph summary of spending behaviour.
2) Five actionable recommendations to reduce unnecessary expenses & improve savings.
3) Suggested monthly budget percentages by category.
4) Anything suspicious to review.
Respond concisely.
"""


    if OPENAI_API_KEY and openai:
        try:
            # use ChatCompletion if available; otherwise fallback
            resp = openai.ChatCompletion.create(
                model="gpt-4o-mini",
                messages=[{"role":"user","content":prompt}],
                max_tokens=600,
                temperature=0.2
            )
            return resp['choices'][0]['message']['content'].strip()
        except Exception as e:
            return f"LLM call failed: {e}\n\nPrompt used:\n{prompt}"
    else:
        # simple rule-based fallback
        recs = [
            "No OpenAI key set — set OPENAI_API_KEY env var for LLM suggestions.",
            "Reduce discretionary category by 10-20% next month.",
            "Set up auto-save: move fixed % of income to savings on payday.",
            "Review / cancel unused subscriptions in Entertainment.",
            "Compare grocery bills weekly & set a target."
        ]
        return "LLM not configured. Suggestions:\n" + "\n".join(f"- {r}" for r in recs)

# ----------------------------
# Streamlit UI
# ----------------------------
st.title("Personal UPI Usage & Financial Analyzer")
st.markdown("Upload UPI PDF statement(s) (or text/CSV). The app will try to parse transactions and provide insights.")

uploaded = st.file_uploader("Upload one or more PDF / CSV / TXT files", type=['pdf','csv','txt'], accept_multiple_files=True)

@st.cache_data
def process_files_streamlit(files):
    dfs = []
    for f in files:
        try:
            content = f.read()
            name = f.name
            if name.lower().endswith(".pdf") or getattr(f, "type", "") == "application/pdf":
                df = parse_pdf_to_df_from_bytes(content)
                if not df.empty:
                    df['source_file'] = name
                    dfs.append(df)
            elif name.lower().endswith(".csv") or getattr(f, "type","").startswith("text/csv"):
                try:
                    df = pd.read_csv(BytesIO(content))
                    dfs.append(df)
                except Exception as e:
                    st.warning(f"CSV read failed for {name}: {e}")
            else:
                # txt fallback
                try:
                    raw = content.decode(errors='ignore')
                except Exception:
                    raw = str(content)
                df = parse_table_like_text(raw)
                if not df.empty:
                    df['source_file'] = name
                    dfs.append(df)
        except Exception as e:
            st.error(f"Failed to process {f.name}: {e}")
    if dfs:
        out = pd.concat(dfs, ignore_index=True, sort=False)
        return out
    else:
        return pd.DataFrame()

if uploaded:
    with st.spinner("Parsing uploaded file(s)..."):
        df = process_files_streamlit(uploaded)
    if df.empty:
        st.error("Parsing produced no transactions. Try different statement export (CSV or text) or check sample format.")
    else:
        st.subheader("Parsed transactions (preview)")
        st.dataframe(df.head(200))

        # Summary metrics
        st.subheader("Summary")
        total_income = df[df['amount_signed'] > 0]['amount'].sum()
        total_spent = df[df['amount_signed'] < 0]['amount'].abs().sum()
        col1, col2 = st.columns(2)
        col1.metric("Total Income (sum credits)", f"{total_income:.2f}")
        col2.metric("Total Spending (sum debits)", f"{total_spent:.2f}")

        # Category breakdown (pie)
        st.subheader("Category breakdown")
        cat = df.groupby('category')['amount'].sum().abs().reset_index().sort_values(by='amount', ascending=False)
        if not cat.empty:
            fig = px.pie(cat, names='category', values='amount', title="Spending by category")
            st.plotly_chart(fig, use_container_width=True)

        # Monthly timeseries
        st.subheader("Spending over time")
        df_ts = df.copy()
        # make sure datetime column is proper
        df_ts['datetime'] = pd.to_datetime(df_ts['datetime'], errors='coerce')
        df_ts['month'] = df_ts['datetime'].dt.to_period("M").astype(str)
        monthly = df_ts[df_ts['amount_signed']<0].groupby('month')['amount_signed'].sum().abs().reset_index().sort_values('month')
        if not monthly.empty:
            fig2 = px.bar(monthly, x='month', y='amount_signed', labels={'amount_signed':'amount'}, title="Monthly spending")
            st.plotly_chart(fig2, use_container_width=True)
        else:
            st.info("No monthly spending data to show (datetime parsing may have failed).")

        # Top merchants
        st.subheader("Top merchants (by spend)")
        topm = df[df['amount_signed']<0].groupby('merchant')['amount_signed'].sum().abs().reset_index().sort_values(by='amount_signed', ascending=False).head(10)
        if not topm.empty:
            st.table(topm)
        else:
            st.info("No merchant spend data found.")

        # Download parsed CSV
        csv = df.to_csv(index=False).encode()
        b64 = base64.b64encode(csv).decode()
        href = f'<a href="data:file/csv;base64,{b64}" download="transactions_parsed.csv">Download parsed CSV</a>'
        st.markdown(href, unsafe_allow_html=True)

        # Filters
        st.subheader("Filter transactions")
        c1, c2 = st.columns([2,1])
        with c1:
            merchant_filter = st.text_input("Merchant contains")
        with c2:
            cat_options = ["All"] + sorted(df['category'].dropna().unique().tolist())
            cat_filter = st.selectbox("Category", options=cat_options)
        filtered = df.copy()
        if merchant_filter:
            filtered = filtered[filtered['merchant'].str.contains(merchant_filter, case=False, na=False)]
        if cat_filter and cat_filter != "All":
            filtered = filtered[filtered['category'] == cat_filter]
        st.dataframe(filtered.head(200))

        # LLM Recommendations
        st.subheader("Personalized recommendations (LLM)")
        if st.button("Get LLM Recommendations"):
            with st.spinner("Generating recommendations..."):
                rec = get_recommendations(df)
                st.markdown("**LLM Recommendations**")
                st.write(rec)
else:
    st.info("Upload your PDF statement(s) or a CSV/text export. Example row format:\n\n`02-09-2024 19:02 17734 debit Flipkart Payment related to Flipkart UTR100000 Shopping`")

# ----------------------------
# End
# ----------------------------
```
